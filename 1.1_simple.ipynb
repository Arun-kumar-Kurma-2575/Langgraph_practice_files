```python
from langgraph.graph import StateGraph,START,END
from typing import TypedDict
```

```python
#defining the state of the graph
class art_state(TypedDict):
    a:int
    b:int
    add:int
    sub:int
    mul:int
    div:float


```

```python
# defining the graph
graph=StateGraph(art_state)
```

```python
def arthimatic(state:art_state):
    a=state['a']
    b=state['b']
    add=a+b
    sub=a-b
    mul=a*b
    div=a/b

    return {'add':add,'sub':sub,"mul":mul,"div":div}
```

```python
# defining the nodes and edges of the graph
graph.add_node('arthimatic',arthimatic)

graph.add_edge(START,'arthimatic')
graph.add_edge('arthimatic',END)
```

```python
workflow=graph.compile()
workflow
```

```python
output=workflow.invoke({'a':10,'b':5})
print(output)
```

```python
state=art_state(**output)
```

```python
print(f' addition of {state["a"]},{state["b"]} is {state["add"]}')
print(f' sub of {state["a"]},{state["b"]} is {state["sub"]}')
print(f' mul of {state["a"]},{state["b"]} is {state["mul"]}')
print(f' div of {state["a"]},{state["b"]} is {state["div"]}')
```
